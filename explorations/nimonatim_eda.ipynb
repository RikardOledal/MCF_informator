# Explore geography API
requiremenst:
- 1 request per second
- Headers w email
- No frequent searches for same search

```python
import httpx

HEADERS = {"User-Agent": "EDA-skyddsrum/0.1 (anja.scherwall@stud.sti.se)"}

def search_address(address):
    response = httpx.get(
        "https://nominatim.openstreetmap.org/search",
        params = {"q": address, "format": "json", "limit": 1},
        headers=HEADERS
    )
    data = response.json()
    return data
```

```python
# json dump to save result and avoid unnecesary calls (from documentation for API)
import json, time

address = ["Storgatan 1, Stockholm", "Provröret 15, Stockholm"]

all = []
for a in address:
    all.append(search_address(a))
    time.sleep(2) # only allowed 1 request/sec

with open("nominatim_test.json", "w", encoding="utf-8") as f:
    json.dump(all, f, ensure_ascii=False, indent=2)
```

## Explore reverse function (lon and lat -> address)

```python
def search_coords(lat, lon):
    response = httpx.get(
        "https://nominatim.openstreetmap.org/reverse",
        params = {"lat": lat, "lon": lon, "format": "json"},
        headers=HEADERS
    )
    response.raise_for_status()
    return response.json()
```

```python
coordinates = [(59.30612537, 17.98231774), (61.83056, 16.09250)]

all_coords = []
for lat, lon in coordinates:
    all_coords.append(search_coords(lat, lon))
    time.sleep(2) # only allowed 1 request/sec

with open("nominatim_reverse_test.json", "w", encoding="utf-8") as f:
    json.dump(all_coords, f, ensure_ascii=False, indent=2)
```

```python
# use address from user and get lat, lon for MCF API
def input_address(address):
    data = search_address(address)

    if not data:                     # empty list
        return None, None

    lat = data[0]["lat"]
    lon = data[0]["lon"]
    return float(lat), float(lon)
```
