# Testing to scrape a page (1 credit)

```python
import os
from dotenv import load_dotenv
from firecrawl import Firecrawl

load_dotenv()

FIRECRAWL_API = os.environ["FIRECRAWL_API_KEY"]

app = Firecrawl()

data = app.scrape(
    "https://www.mcf.se/sv/rad-till-privatpersoner/hemberedskap---preppa-for-en-vecka/grunder-i-hemberedskap/",
    only_main_content=True,
    max_age=172800000,
    parsers=["pdf"],
    formats=["markdown"],
)
```

# Map the site (1 credit)

```python
map_result = app.map(
    "https://www.mcf.se/sv/rad-till-privatpersoner/",
    limit=200,
    include_subdomains=False,
    sitemap="include"
)
```

### Unpack url and make it a list

```python
all_urls = []

for link in map_result.links:
    all_urls.append(link.url)
```

```python
for url in all_urls:
    print(url)
```
